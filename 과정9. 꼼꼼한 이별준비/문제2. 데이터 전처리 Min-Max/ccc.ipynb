```python
import numpy as np
import pandas as pd

with open('전복데이터/abalone_attributes.txt', encoding='utf-8') as f:
    columns = [line.strip() for line in f if line.strip()]

df = pd.read_csv('전복데이터/abalone.txt', header=None, names=columns)

df.head()
```

```python
# 데이터 전처리 Min-Max
label = df['Sex']
numeric_abalone = df.drop(columns=['Sex'])

from sklearn.preprocessing import MinMaxScaler

df_MMscaler = MinMaxScaler()
scaled_data = df_MMscaler.fit_transform(numeric_abalone)
scaled_df = pd.DataFrame(scaled_data, columns=numeric_abalone.columns)

print(scaled_df.head())

# 데이터 전처리 공식으로 직접 구현
function = lambda x: (x - x.min()) / (x.max() - x.min())

scaled_data2 = numeric_abalone.apply(function)

scaled_df2 = pd.DataFrame(scaled_data2, columns=numeric_abalone.columns)

print(scaled_df2.head())

```

```python
# 보너스 과제
# 데이터 전처리 Standardization
from sklearn.preprocessing import StandardScaler
df_Stdscaler = StandardScaler()
scaled_data3 = df_Stdscaler.fit_transform(numeric_abalone)
scaled_df3 = pd.DataFrame(scaled_data3, columns=numeric_abalone.columns)
print(scaled_df3.head())
```

```python
# 데이터 전처리 sampling
from collections import Counter
labels = df['Sex']
data = df.drop(columns=['Sex'])

X_res, y_res = data, labels
print('original dataset shape %s' % Counter(labels))

from imblearn.over_sampling import RandomOverSampler
ros = RandomOverSampler(random_state=42)
X_res, y_res = ros.fit_resample(data, labels)
print('resampled dataset shape %s' % Counter(y_res))    

from imblearn.under_sampling import RandomUnderSampler
rus = RandomUnderSampler(random_state=42)
X_res, y_res = rus.fit_resample(data, labels)
print('resampled dataset shape %s' % Counter(y_res))

# SMOTE 사용
from imblearn.over_sampling import SMOTE
smote = SMOTE(random_state=42)
X_res, y_res = smote.fit_resample(data, labels)
print('resampled dataset shape %s' % Counter(y_res))
```
